# Low-Level and Referential Arrays

> [!info] Definitions & Properties
>  An array is a contiguous collection of memory cells used to store related variables in sequence. In Python, sequence types are implemented either as referential arrays (which store memory addresses referencing objects elsewhere in memory) or compact arrays (which store the raw bit representations of elements directly in contiguous bytes).
> 
>   

## 1. Low-Level Memory Architecture

Computer memory consists of binary bits grouped into 8-bit units called bytes. Each byte is identified by a unique integer called its memory address.

### Random Access Memory (RAM)

Hardware architecture permits access to any memory byte in $O(1)$ time based directly on its memory address.  

### Array Memory Address Calculation

If an array starts at base memory address $\text{start}$ and each cell occupies $\text{cellsize}$ bytes, the memory address of index $i$ is computed in $O(1)$ arithmetic:

$$\text{Address}(A[i]) = \text{start} + (\text{cellsize} \cdot i)$$

## 2. Referential Arrays

Python lists and tuples do not store data values directly inside the array structure. Instead, they store a contiguous sequence of fixed-length memory addresses (object references) pointing to the actual objects.

- On typical 64-bit systems, each address cell uses 8 bytes (64 bits) regardless of the size or type of the object referenced.
    
- Enables lists to store heterogeneous elements and variable-length data (such as strings of arbitrary size).

### Slicing and Shallow Copying

Slicing a sequence (e.g., `temp = primes[3:6]`) constructs a brand-new list instance. However, the cells of the new list point to the exact same element instances as the original list (shallow copy).

```
primes = [2, 3, 5, 7, 11, 13, 17, 19]

temp = primes[3:6]

print(temp)

temp[2]= 15

print(primes)
```

```
primes = [2, 3, 5, 7, 11, 13, 17, 19]
temp = primes[3:6]  # references elements 7, 11, 13

print(temp)

# reassigning an index in temp redirects the pointer to a new object
temp[2] = 15  # primes remains unchanged because integers are immutable

print(primes)
```

### Reference Aliasing and Repetition

Initializing a list using multiplication (e.g., `counters = [0] * 8`) creates 8 cells that all point to the single integer object `0` in memory.

Reassigning an element creates a new integer instance and binds only that specific index to the new address:


```
counters = [0] * 8  # all 8 cells reference id(0)
print(counters)

# immutable update reassigns only index 2
counters[2] += 1  # creates integer 1, assigns reference to counters[2]
print(counters)
```

### The `extend` Method

The `extend` method appends references to existing objects from another sequence to the end of a list without creating copies of those objects.


> [!NOTE] Figure 5.9
> 
>   
> ![[Pasted image 20260923220146.png]]
> 
>   

```
primes = [2, 3, 5, 7, 11, 13, 17, 19]
extras = [23, 29, 31]

primes.extend(extras)  # primes now contains references to objects in extras
print(primes)
```

## 3. Compact Arrays

Compact arrays store raw data directly as contiguous bytes rather than using separate pointer references. Text strings in Python are primary examples of compact arrays.

### Performance and Memory Comparison

|**Feature**|**Referential Array (list, tuple)**|**Compact Array (str, array)**|
|---|---|---|
|**Cell Content**|64-bit memory address|Primary binary data|
|**Memory Consumption**|High (64-bit pointer + object headers)|Low (only the raw bytes for the data type)|
|**Data Types**|Heterogeneous objects allowed|Homogeneous primitive types only|
|**CPU Cache Locality**|Poor (elements scattered across heap)|Excellent (contiguous memory blocks)|

## 4. Python `array` Module

The standard library `array` module provides compact storage for homogeneous primitive C data types.

### Table 5.1: Type Codes Supported by the `array` Module

|**Code**|**C Data Type**|**Typical Size (Bytes)**|
|---|---|---|
|`'b'`|signed char|1|
|`'B'`|unsigned char|1|
|`'u'`|Unicode char|2 or 4|
|`'h'`|signed short int|2|
|`'H'`|unsigned short int|2|
|`'i'`|signed int|2 or 4|
|`'I'`|unsigned int|2 or 4|
|`'l'`|signed long int|4|
|`'L'`|unsigned long int|4|
|`'f'`|float|4|
|`'d'`|float (double precision)|8|

### Practical Memory Footprint Comparison

```
import sys
from array import array

# create identical sequences of 1000 integers
num_list = list(range(1000))
compact_arr = array('i', range(1000))

# memory consumption comparison
print("List reference array size (bytes):", sys.getsizeof(num_list))
print("Compact array buffer size (bytes):", sys.getsizeof(compact_arr))
```

> [!note] Low-Level Custom Arrays
>  The `array` module does not support user-defined composite types. Creating compact arrays of arbitrary structs requires the lower-level `ctypes` module (see Section 5.3.1).
